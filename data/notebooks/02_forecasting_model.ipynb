```python
import pandas as pd
import numpy as np
```

```python
df_reliable = pd.read_pickle("../processed/df_reliable.pkl")

print(df_reliable.shape)
print(df_reliable["report_date"].min())
print(df_reliable["report_date"].max())
```

```python
DAILY_KEYS = [
    "market_name",
    "commodity_name",
    "report_date"
]

df_daily = (
    df_reliable
    .groupby(DAILY_KEYS, as_index=False)
    .agg(
        modal_price=("modal_price", "median"),
        min_price=("min_price", "median"),
        max_price=("max_price", "median"),
        arrivals=("arrivals", "sum"),
        variety_count=("variety", "nunique"),
        grade_count=("grade", "nunique")
    )
)

df_daily = (
    df_daily
    .sort_values(DAILY_KEYS)
    .reset_index(drop=True)
)

print("Original rows:", len(df_reliable))
print("Daily rows:", len(df_daily))
print("Date range:", df_daily["report_date"].min(), "→", df_daily["report_date"].max())
```

```python
df_reliable
```

```python
GROUP_KEYS = [
    "market_name",
    "commodity_name"
]

grp = df_daily.groupby(GROUP_KEYS)["modal_price"]

# Calendar features
df_daily["year"] = df_daily["report_date"].dt.year
df_daily["month"] = df_daily["report_date"].dt.month
df_daily["day_of_week"] = df_daily["report_date"].dt.dayofweek
df_daily["week_of_year"] = (
    df_daily["report_date"].dt.isocalendar().week.astype(int)
)

# Gap since previous observation
df_daily["days_since_last_report"] = (
    df_daily
    .groupby(GROUP_KEYS)["report_date"]
    .diff()
    .dt.days
)

# Price lags
df_daily["price_lag_1"] = grp.shift(1)
df_daily["price_lag_2"] = grp.shift(2)
df_daily["price_lag_3"] = grp.shift(3)
df_daily["price_lag_7"] = grp.shift(7)

# Rolling features — use shift(1) to avoid leakage
df_daily["price_roll_mean_7"] = (
    df_daily
    .groupby(GROUP_KEYS)["modal_price"]
    .transform(
        lambda s: s.shift(1).rolling(7, min_periods=3).mean()
    )
)

df_daily["price_roll_std_7"] = (
    df_daily
    .groupby(GROUP_KEYS)["modal_price"]
    .transform(
        lambda s: s.shift(1).rolling(7, min_periods=3).std()
    )
)

df_daily["price_roll_mean_30"] = (
    df_daily
    .groupby(GROUP_KEYS)["modal_price"]
    .transform(
        lambda s: s.shift(1).rolling(30, min_periods=5).mean()
    )
)

print("Shape:", df_daily.shape)

print(
    df_daily[
        [
            "market_name",
            "commodity_name",
            "report_date",
            "modal_price",
            "price_lag_1",
            "price_lag_7",
            "price_roll_mean_7",
            "price_roll_std_7",
            "price_roll_mean_30"
        ]
    ].head(15)
)
```

```python
# 7th future observation for the same market + commodity

df_daily["target_price"] = (
    df_daily
    .groupby(GROUP_KEYS)["modal_price"]
    .shift(-7)
)

df_daily["target_date"] = (
    df_daily
    .groupby(GROUP_KEYS)["report_date"]
    .shift(-7)
)

df_daily["horizon_days"] = (
    df_daily["target_date"]
    - df_daily["report_date"]
).dt.days

print(
    df_daily[
        [
            "market_name",
            "commodity_name",
            "report_date",
            "modal_price",
            "target_price",
            "target_date",
            "horizon_days"
        ]
    ].head(15)
)
```

```python
# Keep only forecasts where the 7th future observation
# occurs 7 to 14 calendar days later.

df_model = df_daily[
    df_daily["horizon_days"].between(7, 14)
].copy()

# Remove rows where a future target does not exist
df_model = df_model.dropna(
    subset=["target_price"]
).copy()

print("Final modeling rows:", len(df_model))

print("\nHorizon distribution:")
print(
    df_model["horizon_days"]
    .value_counts()
    .sort_index()
)

print("\nDate range:")
print(
    df_model["report_date"].min(),
    "→",
    df_model["report_date"].max()
)
```

```python
# Percentage change from current price to the 7th future observation

df_model["pct_change_7"] = (
    (df_model["target_price"] - df_model["modal_price"])
    / df_model["modal_price"]
)

# Direction classes
df_model["direction"] = pd.cut(
    df_model["pct_change_7"],
    bins=[-np.inf, -0.02, 0.02, np.inf],
    labels=["down", "flat", "up"]
)

print("Direction counts:")
print(df_model["direction"].value_counts())

print("\nDirection percentages:")
print(
    (
        df_model["direction"]
        .value_counts(normalize=True)
        * 100
    ).round(2)
)
```

```python
FEATURES = [
    "market_name",
    "commodity_name",

    "modal_price",
    "min_price",
    "max_price",
    "arrivals",

    "days_since_last_report",

    "price_lag_1",
    "price_lag_2",
    "price_lag_3",
    "price_lag_7",

    "price_roll_mean_7",
    "price_roll_std_7",
    "price_roll_mean_30",

    "year",
    "month",
    "day_of_week",
    "week_of_year"
]

CATEGORICAL = [
    "market_name",
    "commodity_name"
]

print("Number of features:", len(FEATURES))
print("\nFeatures:")
for feature in FEATURES:
    print("-", feature)
```

```python
# Chronological split
# Train:      up to 2025-06-30
# Validation: 2025-07-01 to 2025-12-31
# Test:       2026-01-01 onward

split_train_end = pd.Timestamp("2025-06-30")
split_val_end = pd.Timestamp("2025-12-31")

train = df_model[
    df_model["report_date"] <= split_train_end
].copy()

val = df_model[
    (df_model["report_date"] > split_train_end) &
    (df_model["report_date"] <= split_val_end)
].copy()

test = df_model[
    df_model["report_date"] > split_val_end
].copy()

print("Train shape:", train.shape)
print("Validation shape:", val.shape)
print("Test shape:", test.shape)

print("\nDate ranges:")

print(
    "Train:",
    train["report_date"].min(),
    "→",
    train["report_date"].max()
)

print(
    "Validation:",
    val["report_date"].min(),
    "→",
    val["report_date"].max()
)

print(
    "Test:",
    test["report_date"].min(),
    "→",
    test["report_date"].max()
)
```

```python
for col in CATEGORICAL:
    categories = df_model[col].astype("category").cat.categories

    train[col] = pd.Categorical(
        train[col],
        categories=categories
    )

    val[col] = pd.Categorical(
        val[col],
        categories=categories
    )

    test[col] = pd.Categorical(
        test[col],
        categories=categories
    )

print("Categorical columns prepared:")
print(CATEGORICAL)

print("\nTrain dtypes:")
print(train[CATEGORICAL].dtypes)
```

```python
X_train_reg = train[FEATURES]
X_val_reg = val[FEATURES]
X_test_reg = test[FEATURES]

y_train_reg = train["target_price"]
y_val_reg = val["target_price"]
y_test_reg = test["target_price"]

print("X_train:", X_train_reg.shape)
print("X_val:", X_val_reg.shape)
print("X_test:", X_test_reg.shape)

print("\ny_train:", y_train_reg.shape)
print("y_val:", y_val_reg.shape)
print("y_test:", y_test_reg.shape)
```

```python
from lightgbm import LGBMRegressor, early_stopping, log_evaluation

regressor = LGBMRegressor(
    objective="regression",
    n_estimators=3000,
    learning_rate=0.03,
    num_leaves=63,
    min_data_in_leaf=100,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1
)

regressor.fit(
    X_train_reg,
    np.log1p(y_train_reg),

    eval_set=[
        (X_val_reg, np.log1p(y_val_reg))
    ],

    eval_metric="l2",

    categorical_feature=CATEGORICAL,

    callbacks=[
        early_stopping(100),
        log_evaluation(100)
    ]
)
```

```python
# Predict log(price), then convert back to the original price scale

pred_log = regressor.predict(
    X_test_reg
)

pred_price = np.expm1(pred_log)

# Prices cannot be negative
pred_price = np.maximum(pred_price, 0)

print("Number of predictions:", len(pred_price))
print("First 10 predictions:")
print(pred_price[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

mae = mean_absolute_error(
    y_test_reg,
    pred_price
)

rmse = np.sqrt(
    mean_squared_error(
        y_test_reg,
        pred_price
    )
)

smape = (
    100
    * np.mean(
        2 * np.abs(pred_price - y_test_reg)
        /
        (
            np.abs(y_test_reg)
            + np.abs(pred_price)
            + 1e-8
        )
    )
)

print(f"MAE   : ₹{mae:,.2f}")
print(f"RMSE  : ₹{rmse:,.2f}")
print(f"sMAPE : {smape:.2f}%")
```

```python
# Naive forecast:
# assume the future price will be the current modal price

naive_pred = test["modal_price"].values

naive_mae = mean_absolute_error(
    y_test_reg,
    naive_pred
)

naive_rmse = np.sqrt(
    mean_squared_error(
        y_test_reg,
        naive_pred
    )
)

naive_smape = (
    100
    * np.mean(
        2 * np.abs(naive_pred - y_test_reg)
        /
        (
            np.abs(y_test_reg)
            + np.abs(naive_pred)
            + 1e-8
        )
    )
)

print("NAIVE BASELINE")
print(f"MAE   : ₹{naive_mae:,.2f}")
print(f"RMSE  : ₹{naive_rmse:,.2f}")
print(f"sMAPE : {naive_smape:.2f}%")

print("\nLIGHTGBM")
print(f"MAE   : ₹{mae:,.2f}")
print(f"RMSE  : ₹{rmse:,.2f}")
print(f"sMAPE : {smape:.2f}%")
```

```python
# Recalculate test predictions and metrics from scratch
# to make sure no variable was accidentally overwritten.

pred_log_check = regressor.predict(X_test_reg)
pred_price_check = np.maximum(np.expm1(pred_log_check), 0)

actual = y_test_reg.to_numpy()
naive_check = test["modal_price"].to_numpy()

# LightGBM
lgbm_mae_check = mean_absolute_error(actual, pred_price_check)
lgbm_rmse_check = np.sqrt(
    mean_squared_error(actual, pred_price_check)
)

lgbm_smape_check = (
    100
    * np.mean(
        2 * np.abs(pred_price_check - actual)
        / (np.abs(actual) + np.abs(pred_price_check) + 1e-8)
    )
)

# Naive
naive_mae_check = mean_absolute_error(actual, naive_check)
naive_rmse_check = np.sqrt(
    mean_squared_error(actual, naive_check)
)

naive_smape_check = (
    100
    * np.mean(
        2 * np.abs(naive_check - actual)
        / (np.abs(actual) + np.abs(naive_check) + 1e-8)
    )
)

print("LIGHTGBM")
print(f"MAE   : ₹{lgbm_mae_check:,.2f}")
print(f"RMSE  : ₹{lgbm_rmse_check:,.2f}")
print(f"sMAPE : {lgbm_smape_check:.2f}%")

print("\nNAIVE")
print(f"MAE   : ₹{naive_mae_check:,.2f}")
print(f"RMSE  : ₹{naive_rmse_check:,.2f}")
print(f"sMAPE : {naive_smape_check:.2f}%")
```

```python
# Encode DOWN / FLAT / UP
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

label_encoder.fit(["down", "flat", "up"])

train["direction_encoded"] = label_encoder.transform(
    train["direction"]
)

val["direction_encoded"] = label_encoder.transform(
    val["direction"]
)

test["direction_encoded"] = label_encoder.transform(
    test["direction"]
)

print("Class mapping:")
print(
    dict(
        zip(
            label_encoder.classes_,
            range(len(label_encoder.classes_))
        )
    )
)

print("\nTrain class counts:")
print(train["direction"].value_counts())
```

```python
X_train_cls = train[FEATURES]
X_val_cls = val[FEATURES]
X_test_cls = test[FEATURES]

y_train_cls = train["direction_encoded"]
y_val_cls = val["direction_encoded"]
y_test_cls = test["direction_encoded"]

print("X_train:", X_train_cls.shape)
print("X_val:", X_val_cls.shape)
print("X_test:", X_test_cls.shape)

print("\ny_train:", y_train_cls.shape)
print("y_val:", y_val_cls.shape)
print("y_test:", y_test_cls.shape)
```

```python
from lightgbm import LGBMClassifier, early_stopping, log_evaluation

classifier = LGBMClassifier(
    objective="multiclass",
    num_class=3,
    n_estimators=3000,
    learning_rate=0.05,
    num_leaves=63,
    min_data_in_leaf=200,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1
)

classifier.fit(
    X_train_cls,
    y_train_cls,

    eval_set=[
        (X_val_cls, y_val_cls)
    ],

    eval_metric="multi_logloss",

    categorical_feature=CATEGORICAL,

    callbacks=[
        early_stopping(100),
        log_evaluation(100)
    ]
)
```

```python
# Predict class labels on the test set

pred_direction_encoded = (
    classifier
    .predict(X_test_cls)
    .astype(int)
    .ravel()
)

pred_direction = label_encoder.inverse_transform(
    pred_direction_encoded
)

print("Number of predictions:", len(pred_direction))

print("\nFirst 20 predictions:")
print(pred_direction[:20])
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    y_test_cls,
    pred_direction_encoded
)

print(f"Overall classification accuracy: {accuracy:.4f}")
print(f"Overall classification accuracy: {accuracy * 100:.2f}%")
```

```python
# Create a clean test prediction table

results = test[
    [
        "report_date",
        "target_date",
        "market_name",
        "commodity_name",
        "modal_price",
        "target_price",
        "direction"
    ]
].copy()

# Regression prediction
results["predicted_price"] = pred_price

# Classification prediction
results["predicted_direction"] = pred_direction

# Probability of each class
results["prob_down"] = probabilities[:, 0]
results["prob_flat"] = probabilities[:, 1]
results["prob_up"] = probabilities[:, 2]

# Highest probability = model confidence
results["direction_confidence"] = probabilities.max(axis=1)

print(results.head(10))
```

```python
# Expected percentage change according to the regression model

results["predicted_pct_change"] = (
    (
        results["predicted_price"]
        - results["modal_price"]
    )
    / results["modal_price"]
) * 100

print(
    results[
        [
            "commodity_name",
            "market_name",
            "report_date",
            "modal_price",
            "predicted_price",
            "predicted_pct_change",
            "predicted_direction",
            "direction_confidence"
        ]
    ].head(10)
)
```

```python
def make_recommendation(row):

    if (
        row["predicted_direction"] == "up"
        and row["direction_confidence"] >= 0.60
    ):
        return "HOLD"

    elif (
        row["predicted_direction"] == "down"
        and row["direction_confidence"] >= 0.60
    ):
        return "SELL"

    else:
        return "NO STRONG SIGNAL"


results["recommendation"] = results.apply(
    make_recommendation,
    axis=1
)

print("Recommendation counts:")
print(results["recommendation"].value_counts())

print("\nExamples:")
print(
    results[
        [
            "commodity_name",
            "market_name",
            "modal_price",
            "predicted_price",
            "predicted_pct_change",
            "predicted_direction",
            "direction_confidence",
            "recommendation"
        ]
    ].head(20)
)
```

```python
# Evaluate only the recommendations where the system
# actually made a decision.

decision_mask = results["recommendation"].isin(
    ["HOLD", "SELL"]
)

decision_results = results.loc[
    decision_mask
].copy()

# Actual direction
decision_results["correct_direction"] = (
    decision_results["predicted_direction"]
    == decision_results["direction"]
)

print("Total recommendations:", len(results))
print("HOLD/SELL decisions:", len(decision_results))

print("\nDirection accuracy on HOLD/SELL decisions:")
print(
    decision_results["correct_direction"].mean() * 100
)

print("\nAccuracy by recommendation:")
print(
    decision_results
    .groupby("recommendation")["correct_direction"]
    .mean()
    .mul(100)
)
```

```python
# Check agreement between regression-implied movement
# and classification direction.

results["regression_direction"] = np.select(
    [
        results["predicted_pct_change"] < -2,
        results["predicted_pct_change"] > 2
    ],
    [
        "down",
        "up"
    ],
    default="flat"
)

results["models_agree"] = (
    results["regression_direction"]
    == results["predicted_direction"]
)

print("Model agreement:")
print(
    results["models_agree"].value_counts()
)

print("\nAgreement percentage:")
print(
    results["models_agree"].mean() * 100
)
```

```python
def make_strict_recommendation(row):

    if (
        row["predicted_direction"] == "up"
        and row["regression_direction"] == "up"
        and row["direction_confidence"] >= 0.60
    ):
        return "HOLD"

    elif (
        row["predicted_direction"] == "down"
        and row["regression_direction"] == "down"
        and row["direction_confidence"] >= 0.60
    ):
        return "SELL"

    else:
        return "NO STRONG SIGNAL"


results["strict_recommendation"] = results.apply(
    make_strict_recommendation,
    axis=1
)

print("Strict recommendation counts:")
print(
    results["strict_recommendation"].value_counts()
)
```

```python
# Naive directional baseline:
# predict the future direction from the previous price movement.

naive_direction = np.where(
    test["modal_price"] > test["price_lag_1"],
    "up",
    np.where(
        test["modal_price"] < test["price_lag_1"],
        "down",
        "flat"
    )
)

naive_direction_accuracy = (
    naive_direction == test["direction"].astype(str).values
).mean()

print(
    f"Naive direction accuracy: "
    f"{naive_direction_accuracy * 100:.2f}%"
)

non_flat_mask_naive = (
    test["direction"].astype(str) != "flat"
)

naive_non_flat_accuracy = (
    naive_direction[non_flat_mask_naive]
    == test.loc[non_flat_mask_naive, "direction"].astype(str).values
).mean()

print(
    f"Naive UP/DOWN accuracy: "
    f"{naive_non_flat_accuracy * 100:.2f}%"
)
```

```python
# Direction performance by commodity on the test set

crop_results = []

for crop, group in test.groupby("commodity_name"):

    idx = group.index

    actual = group["direction"].astype(str).values
    predicted = pd.Series(
        pred_direction,
        index=test.index
    ).loc[idx].values

    overall_acc = (
        predicted == actual
    ).mean()

    non_flat = actual != "flat"

    if non_flat.sum() > 0:
        non_flat_acc = (
            predicted[non_flat] == actual[non_flat]
        ).mean()
    else:
        non_flat_acc = np.nan

    crop_results.append({
        "commodity": crop,
        "test_rows": len(group),
        "overall_accuracy": overall_acc,
        "up_down_accuracy": non_flat_acc
    })

per_crop_accuracy = (
    pd.DataFrame(crop_results)
    .sort_values(
        "up_down_accuracy",
        ascending=False
    )
)

print(per_crop_accuracy.head(20))
```

```python
# Evaluate HOLD/SELL recommendations against baselines
# using the original test dataframe for price_lag_1.

issued_mask = results["recommendation"].isin(
    ["HOLD", "SELL"]
)

issued = results.loc[issued_mask].copy()

# Map recommendation to predicted direction
issued["predicted_action_direction"] = issued[
    "recommendation"
].map({
    "HOLD": "up",
    "SELL": "down"
})

# Actual direction
issued["actual_direction"] = issued["direction"].astype(str)

# -----------------------------
# 1. Model accuracy
# -----------------------------

model_action_accuracy = (
    issued["predicted_action_direction"]
    == issued["actual_direction"]
).mean()

# -----------------------------
# 2. Majority-class baseline
# -----------------------------

actual_counts = issued["actual_direction"].value_counts()

majority_direction = actual_counts.index[0]

majority_accuracy = (
    issued["actual_direction"]
    == majority_direction
).mean()

# -----------------------------
# 3. Naive previous-price baseline
# -----------------------------

# Get price_lag_1 from the original test dataframe
issued_lag1 = test.loc[issued.index, "price_lag_1"]

naive_action_direction = np.where(
    issued["modal_price"].values > issued_lag1.values,
    "up",
    np.where(
        issued["modal_price"].values < issued_lag1.values,
        "down",
        "flat"
    )
)

naive_accuracy = (
    naive_action_direction
    == issued["actual_direction"].values
).mean()

# -----------------------------
# Results
# -----------------------------

print("Issued HOLD/SELL decisions:", len(issued))

print("\nActual direction distribution:")
print(actual_counts)

print("\nMODEL")
print(f"Accuracy: {model_action_accuracy * 100:.2f}%")

print("\nMAJORITY BASELINE")
print(
    f"Always predict '{majority_direction}': "
    f"{majority_accuracy * 100:.2f}%"
)

print("\nNAIVE PREVIOUS-PRICE BASELINE")
print(
    f"Accuracy: {naive_accuracy * 100:.2f}%"
)
```

```python
# ------------------------------------------------------------
# SAME-DAY CROSS-MARKET FEATURES
# ------------------------------------------------------------
#
# For each commodity + date:
#   - median price across OTHER markets
#   - mean price across OTHER markets
#   - current market's deviation from other-market median
#   - fraction of OTHER markets that moved up
#   - fraction of OTHER markets that moved down
#
# These are calculated before target creation and use only
# information available on the current report date.
# ------------------------------------------------------------

cross = df_daily[
    [
        "market_name",
        "commodity_name",
        "report_date",
        "modal_price"
    ]
].copy()

# Previous price for each market + commodity
cross["prev_price"] = (
    cross
    .groupby(["market_name", "commodity_name"])["modal_price"]
    .shift(1)
)

# Same-day commodity statistics across all markets
cross["same_day_market_count"] = (
    cross
    .groupby(["commodity_name", "report_date"])["market_name"]
    .transform("nunique")
)

cross["same_day_price_sum"] = (
    cross
    .groupby(["commodity_name", "report_date"])["modal_price"]
    .transform("sum")
)

cross["same_day_price_mean_all"] = (
    cross
    .groupby(["commodity_name", "report_date"])["modal_price"]
    .transform("mean")
)

cross["same_day_price_median_all"] = (
    cross
    .groupby(["commodity_name", "report_date"])["modal_price"]
    .transform("median")
)

# Exclude the current market from the mean
cross["other_market_count"] = (
    cross["same_day_market_count"] - 1
)

cross["cross_market_mean"] = np.where(
    cross["other_market_count"] > 0,
    (
        cross["same_day_price_sum"]
        - cross["modal_price"]
    ) / cross["other_market_count"],
    np.nan
)

# Current market relative to same-day statewide median
cross["price_vs_market_median"] = (
    cross["modal_price"]
    / cross["same_day_price_median_all"]
)

cross["price_vs_other_market_mean"] = (
    cross["modal_price"]
    / cross["cross_market_mean"]
)

# Current market's own previous movement
cross["current_market_pct_change"] = (
    (cross["modal_price"] - cross["prev_price"])
    / cross["prev_price"]
)

# Count markets moving up/down on that date
cross["markets_up_count"] = (
    cross
    .assign(
        market_up=(
            cross["modal_price"]
            > cross["prev_price"]
        )
    )
    .groupby(["commodity_name", "report_date"])["market_up"]
    .transform("sum")
)

cross["markets_down_count"] = (
    cross
    .assign(
        market_down=(
            cross["modal_price"]
            < cross["prev_price"]
        )
    )
    .groupby(["commodity_name", "report_date"])["market_down"]
    .transform("sum")
)

cross["pct_markets_up"] = (
    cross["markets_up_count"]
    / cross["same_day_market_count"]
)

cross["pct_markets_down"] = (
    cross["markets_down_count"]
    / cross["same_day_market_count"]
)

# Keep only the features we want
CROSS_FEATURES = [
    "same_day_market_count",
    "cross_market_mean",
    "same_day_price_median_all",
    "price_vs_market_median",
    "price_vs_other_market_mean",
    "pct_markets_up",
    "pct_markets_down"
]

# Merge back into df_daily
df_daily = df_daily.merge(
    cross[
        [
            "market_name",
            "commodity_name",
            "report_date"
        ] + CROSS_FEATURES
    ],
    on=[
        "market_name",
        "commodity_name",
        "report_date"
    ],
    how="left"
)

print("Cross-market features added.")

print(
    df_daily[
        [
            "market_name",
            "commodity_name",
            "report_date",
            "modal_price"
        ] + CROSS_FEATURES
    ].head(20)
)
```

```python
# Recalculate same-day market movement using OTHER markets only

cross_move = df_daily[
    [
        "market_name",
        "commodity_name",
        "report_date",
        "modal_price"
    ]
].copy()

# Previous reported price for each market + commodity
cross_move["prev_price"] = (
    cross_move
    .groupby(["market_name", "commodity_name"])["modal_price"]
    .shift(1)
)

# Current market movement
cross_move["current_up"] = (
    cross_move["modal_price"]
    > cross_move["prev_price"]
)

cross_move["current_down"] = (
    cross_move["modal_price"]
    < cross_move["prev_price"]
)

# Total markets moving up/down for commodity + date
total_up = (
    cross_move
    .groupby(["commodity_name", "report_date"])["current_up"]
    .transform("sum")
)

total_down = (
    cross_move
    .groupby(["commodity_name", "report_date"])["current_down"]
    .transform("sum")
)

market_count = (
    cross_move
    .groupby(["commodity_name", "report_date"])["market_name"]
    .transform("nunique")
)

# Exclude the current market
other_market_count = market_count - 1

other_up = total_up - cross_move["current_up"].astype(int)
other_down = total_down - cross_move["current_down"].astype(int)

df_daily["pct_other_markets_up"] = np.where(
    other_market_count > 0,
    other_up / other_market_count,
    np.nan
)

df_daily["pct_other_markets_down"] = np.where(
    other_market_count > 0,
    other_down / other_market_count,
    np.nan
)

print(
    df_daily[
        [
            "market_name",
            "commodity_name",
            "report_date",
            "same_day_market_count",
            "pct_other_markets_up",
            "pct_other_markets_down"
        ]
    ].head(20)
)

print("\nMissing values:")
print(
    df_daily[
        [
            "pct_other_markets_up",
            "pct_other_markets_down"
        ]
    ].isna().sum()
)
```
